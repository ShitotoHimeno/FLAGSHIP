## YOLO使ってみた

```python
from ultralytics import YOLO

model_path = "yolov5s.pt"

model = YOLO(model_path)

print(model)
```

```python
results = model("/home/jovyan/work/data/IMG_0562.jpg", device = 'cpu')

for r in results:
    r.show()
    print(r.boxes)
```

```python
def detect_objects_in_image(model, image_path):
    results = model(image_path)
    return results

# 使用例
image_path = '/home/jovyan/work/data/bus.jpg'
results = detect_objects_in_image(model, image_path)
print(f"検出された物体: {len(results[0].boxes)} 個")
```
